```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}
def print_graph(graf):
  for node in graf:
    print(f"{node} -> {graf[node]}")

print_graph(graph)
```

```python
from collections import deque

def bfs(graph, start):
  visited = set()
  queue =deque([start])

  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      print(vertex, end=" ")
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

bfs(graph, 'A')
```

```python
def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()

  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)

dfs(graph, 'A')
```

```python
graph = {
    '1': ['2', '3'],
    '2': ['1', '3', '4'],
    '3': ['1', '2', '4', '5'],
    '4': ['2', '3', '5'],
    '5': ['3', '4']
}
def print_graph(graf):
  for node in graf:
    print(f"{node} -> {graf[node]}")

print_graph(graph)
```

```python
from collections import deque

def bfs(graph, start):
  visited = set()
  queue =deque([start])

  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      print(vertex, end=" ")
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

print("BFS: ", end="")
bfs(graph, '1')

def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()

  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)

print("\nDFS: ",end= "")
dfs(graph, '2')
```

```python
def bfs_return_list(graph, start):
    visited = set()
    queue = deque([start])
    order = []
    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            order.append(vertex)
            visited.add(vertex)
            for neighbor in graph[vertex]:
                if neighbor not in visited:
                    queue.append(neighbor)
    return order

print("List urutan kunjungan BFS:", bfs_return_list(graph, '1'))
```

```python
def find_path(graph, start, end, path=None):
    if path is None:
        path = []
    path = path + [start]
    if start == end:
        return path
    if start not in graph:
        return None
    for neighbor in graph[start]:
        if neighbor not in path:
            new_path = find_path(graph, neighbor, end, path)
            if new_path:
                return new_path
    return None

print("Jalur 1 ke 5:", find_path(graph, '1', '5'))
print("Jalur 2 ke 5:", find_path(graph, '2', '5'))
```

```python
def is_connected(graph):
    if not graph:
        return True
    start = next(iter(graph))
    visited = set()

    def dfs_check(node):
        visited.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                dfs_check(neighbor)

    dfs_check(start)
    return len(visited) == len(graph)

print("Apakah graf terhubung?", is_connected(graph))
```
